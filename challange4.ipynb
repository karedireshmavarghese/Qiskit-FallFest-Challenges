```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
```

```python
# Part A: Correlated Outcomes

qc_A = QuantumCircuit(2, 2)

# Put qubit 0 into superposition
qc_A.h(0)

# Entangle qubit 0 and qubit 1
qc_A.cx(0, 1)

# Measure both qubits
qc_A.measure([0, 1], [0, 1])

# Display circuit
print("Part A Circuit:")
print(qc_A.draw())

# Run for 100 shots
simulator = AerSimulator()
result_A = simulator.run(qc_A, shots=100).result()

counts_A = result_A.get_counts()

print("Part A Results:")
print(counts_A)
```

```python
plot_histogram(
    counts_A,
    title="Part A - Correlated Outcomes"
)
plt.show()
```

```python
# Part B: Opposite Outcomes

qc_B = QuantumCircuit(2, 2)

# Create the same entangled state
qc_B.h(0)
qc_B.cx(0, 1)

# Flip qubit 1
qc_B.x(1)

# Measure both qubits
qc_B.measure([0, 1], [0, 1])

# Display circuit
print("Part B Circuit:")
print(qc_B.draw())

# Run for 100 shots
result_B = simulator.run(qc_B, shots=100).result()

counts_B = result_B.get_counts()

print("Part B Results:")
print(counts_B)
```

```python
plot_histogram(
    counts_B,
    title="Part B - Opposite Outcomes"
)
plt.show()
```

```python
plot_histogram(
    [counts_A, counts_B],
    legend=["Part A - Correlated", "Part B - Opposite"],
    title="Comparison of Correlated and Opposite Outcomes"
)
plt.show()
```
